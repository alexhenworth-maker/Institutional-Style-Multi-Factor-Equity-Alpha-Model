```python
# ============================================================
# CELL 1 — PROJECT SETUP
# Institutional Multi-Factor Equity Alpha Model
# ============================================================

!pip -q install yfinance pandas numpy scipy statsmodels \
    scikit-learn requests pyarrow plotly seaborn joblib

import os
import time
import json
import math
import warnings
from pathlib import Path
from dataclasses import dataclass
from typing import Dict, List, Optional, Tuple

import numpy as np
import pandas as pd
import requests
import yfinance as yf

import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import plotly.graph_objects as go

from scipy.stats import zscore, spearmanr
from sklearn.linear_model import LinearRegression
from sklearn.covariance import LedoitWolf

warnings.filterwarnings("ignore")

PROJECT_ROOT = Path.cwd() / "multi_factor_equity_alpha"

DATA_DIR = PROJECT_ROOT / "data"
RAW_DIR = DATA_DIR / "raw"
PROCESSED_DIR = DATA_DIR / "processed"
CACHE_DIR = DATA_DIR / "cache"

OUTPUT_DIR = PROJECT_ROOT / "outputs"
CHART_DIR = OUTPUT_DIR / "charts"
TABLE_DIR = OUTPUT_DIR / "tables"

for directory in [
    RAW_DIR,
    PROCESSED_DIR,
    CACHE_DIR,
    CHART_DIR,
    TABLE_DIR
]:
    directory.mkdir(parents=True, exist_ok=True)

print(f"Project directory: {PROJECT_ROOT}")
# ============================================================
# CELL 2 — CONFIGURATION
# ============================================================

@dataclass
class Config:
    start_date: str = "2015-01-01"
    end_date: str = "2025-12-31"

    rebalance_frequency: str = "ME"

    top_quantile: float = 0.20

    transaction_cost_bps: float = 10.0

    winsor_lower: float = 0.01
    winsor_upper: float = 0.99

    min_history_days: int = 252

    sec_delay_seconds: float = 0.15


    sec_user_agent: str = "QuantResearch student@example.com"


CONFIG = Config()

TICKERS = [
    "AAPL", "MSFT", "AMZN", "GOOGL", "META",
    "NVDA", "JPM", "JNJ", "PG", "XOM",
    "UNH", "HD", "MA", "V", "KO",
    "PEP", "COST", "WMT", "CAT", "IBM"
]

FACTOR_WEIGHTS = {
    "momentum": 0.25,
    "value": 0.20,
    "quality": 0.20,
    "growth": 0.15,
    "low_vol": 0.10,
    "size": 0.10,
}

print("Universe:", len(TICKERS))
print("Factor weights:", FACTOR_WEIGHTS)


# ============================================================
# CELL 3 — SEC EDGAR CLIENT
# ============================================================

class SECClient:
    """
    Lightweight SEC Company Facts client.

    Design goals:
        - Respect SEC request identification requirements.
        - Cache responses locally.
        - Handle transient failures.
        - Avoid repeatedly downloading identical data.
    """

    BASE_URL = "https://data.sec.gov"

    def __init__(self, user_agent: str):
        self.session = requests.Session()

        self.session.headers.update({
            "User-Agent": user_agent,
            "Accept-Encoding": "gzip, deflate",
            "Host": "data.sec.gov"
        })

        self.cache_dir = CACHE_DIR / "sec"
        self.cache_dir.mkdir(parents=True, exist_ok=True)

    def _request_json(
        self,
        url: str,
        cache_name: str,
        retries: int = 3
    ) -> dict:

        cache_path = self.cache_dir / cache_name



        if cache_path.exists():
            try:
                with open(cache_path, "r") as f:
                    return json.load(f)
            except Exception:
                pass


        for attempt in range(retries):

            try:
                response = self.session.get(
                    url,
                    timeout=30
                )

                if response.status_code == 200:

                    data = response.json()

                    with open(cache_path, "w") as f:
                        json.dump(data, f)

                    time.sleep(CONFIG.sec_delay_seconds)

                    return data

                elif response.status_code == 429:

                    wait = 2 ** attempt
                    print(
                        f"SEC rate limit encountered. "
                        f"Waiting {wait}s..."
                    )
                    time.sleep(wait)

                else:

                    print(
                        f"SEC request failed: "
                        f"{response.status_code}"
                    )

            except requests.RequestException as exc:

                print(
                    f"SEC request error: {exc}"
                )

                time.sleep(2 ** attempt)

        raise RuntimeError(
            f"Unable to retrieve SEC data: {url}"
        )

    def company_facts(self, cik: str) -> dict:

        cik = str(cik).zfill(10)

        url = (
            f"{self.BASE_URL}/api/xbrl/"
            f"companyfacts/CIK{cik}.json"
        )

        return self._request_json(
            url=url,
            cache_name=f"{cik}_companyfacts.json"
        )


sec = SECClient(CONFIG.sec_user_agent)

print("SEC client initialized.")
# ============================================================
# CELL 4 — SEC TICKER → CIK MAPPING
# ============================================================

def get_sec_ticker_map() -> pd.DataFrame:
    """
    Download SEC's public ticker/CIK mapping.
    """

    url = "https://www.sec.gov/files/company_tickers.json"

    headers = {
        "User-Agent": CONFIG.sec_user_agent
    }

    response = requests.get(
        url,
        headers=headers,
        timeout=30
    )

    response.raise_for_status()

    raw = response.json()

    rows = []

    for item in raw.values():

        rows.append({
            "ticker": item["ticker"].upper(),
            "cik": str(item["cik_str"]).zfill(10),
            "company": item["title"]
        })

    return pd.DataFrame(rows)


sec_tickers = get_sec_ticker_map()

ticker_map = (
    sec_tickers
    .drop_duplicates("ticker")
    .set_index("ticker")
)

ticker_map.head()
# ============================================================
# CELL 5 — MARKET DATA DOWNLOAD
# ============================================================

def download_market_data(
    tickers: List[str],
    start: str,
    end: str
) -> pd.DataFrame:
    """
    Download adjusted historical prices.

    MultiIndex output from yfinance is converted into
    long format for easier factor engineering.
    """

    try:

        data = yf.download(
            tickers=tickers,
            start=start,
            end=end,
            auto_adjust=True,
            progress=False,
            group_by="column",
            threads=True
        )

    except Exception as exc:

        raise RuntimeError(
            f"Yahoo Finance download failed: {exc}"
        )

    if data.empty:
        raise ValueError(
            "No market data returned."
        )



    if isinstance(data.columns, pd.MultiIndex):

        close = data["Close"].copy()

    else:

        close = data[["Close"]].copy()
        close.columns = [tickers[0]]

    close = (
        close
        .replace([np.inf, -np.inf], np.nan)
        .sort_index()
    )

    return close


prices = download_market_data(
    TICKERS,
    CONFIG.start_date,
    CONFIG.end_date
)

print(prices.shape)
prices.tail()


# ============================================================
# CELL 6 — SEC FUNDAMENTAL DATA EXTRACTION
# ============================================================

SEC_TAGS = {
    "revenue": [
        "RevenueFromContractWithCustomerExcludingAssessedTax",
        "Revenues"
    ],

    "net_income": [
        "NetIncomeLoss"
    ],

    "assets": [
        "Assets"
    ],

    "equity": [
        "StockholdersEquity",
        "StockholdersEquityIncludingPortionAttributableToNoncontrollingInterest"
    ],

    "operating_income": [
        "OperatingIncomeLoss"
    ],

    "cash": [
        "CashAndCashEquivalentsAtCarryingValue"
    ],

    "shares": [
        "EntityCommonStockSharesOutstanding"
    ]
}


def extract_sec_concept(
    facts: dict,
    concept_names: List[str],
    ticker: str
) -> pd.DataFrame:
    """
    Extract the first available SEC XBRL concept.

    Each observation retains:
        - period end
        - filing date
        - value
        - form
        - fiscal period
        - fiscal year
    """

    facts_root = facts.get("facts", {})


    namespaces = ["us-gaap", "dei"]

    for namespace in namespaces:

        namespace_data = facts_root.get(namespace, {})

        for concept in concept_names:

            if concept not in namespace_data:
                continue

            concept_data = namespace_data[concept]

            units = concept_data.get("units", {})


            unit_name = (
                "USD"
                if "USD" in units
                else next(iter(units), None)
            )

            if unit_name is None:
                continue

            observations = units[unit_name]

            records = []

            for obs in observations:

                if "val" not in obs:
                    continue

                records.append({
                    "ticker": ticker,
                    "concept": concept,
                    "value": obs.get("val"),
                    "unit": unit_name,
                    "start": obs.get("start"),
                    "end": obs.get("end"),
                    "filed": obs.get("filed"),
                    "form": obs.get("form"),
                    "fp": obs.get("fp"),
                    "fy": obs.get("fy"),
                    "frame": obs.get("frame")
                })

            if records:
                return pd.DataFrame(records)

    return pd.DataFrame()


def download_fundamentals(
    tickers: List[str]
) -> pd.DataFrame:

    all_records = []

    for ticker in tickers:

        try:

            if ticker not in ticker_map.index:
                print(f"{ticker}: no SEC CIK mapping.")
                continue

            cik = ticker_map.loc[ticker, "cik"]

            facts = sec.company_facts(cik)

            for factor_name, tags in SEC_TAGS.items():

                df = extract_sec_concept(
                    facts,
                    tags,
                    ticker
                )

                if df.empty:
                    continue

                df["fundamental"] = factor_name

                all_records.append(df)

        except Exception as exc:

            print(
                f"Fundamental download failed "
                f"for {ticker}: {exc}"
            )

    if not all_records:
        raise RuntimeError(
            "No SEC fundamental data retrieved."
        )

    fundamentals = pd.concat(
        all_records,
        ignore_index=True
    )

    fundamentals["filed"] = pd.to_datetime(
        fundamentals["filed"],
        errors="coerce"
    )

    fundamentals["end"] = pd.to_datetime(
        fundamentals["end"],
        errors="coerce"
    )

    fundamentals["value"] = pd.to_numeric(
        fundamentals["value"],
        errors="coerce"
    )

    fundamentals = (
        fundamentals
        .dropna(subset=["filed", "value"])
        .sort_values(["ticker", "fundamental", "filed"])
    )

    return fundamentals


fundamentals = download_fundamentals(TICKERS)

print(fundamentals.shape)
fundamentals.head()



# ============================================================
# CELL 7 — POINT-IN-TIME FUNDAMENTAL ENGINE
# ============================================================

def point_in_time_fundamentals(
    fundamentals: pd.DataFrame,
    rebalance_dates: pd.DatetimeIndex
) -> pd.DataFrame:
    """
    Build a point-in-time fundamental panel.

    For every rebalance date, only information with:

        filing_date <= rebalance_date

    is eligible.

    When multiple observations exist, the most recently
    filed observation is selected.

    This is critical for avoiding look-ahead bias.
    """

    output = []

    fundamentals = fundamentals.copy()

    fundamentals = fundamentals.sort_values(
        ["ticker", "fundamental", "filed"]
    )

    for rebalance_date in rebalance_dates:

        available = fundamentals[
            fundamentals["filed"] <= rebalance_date
        ]

        if available.empty:
            continue

        latest = (
            available
            .sort_values("filed")
            .groupby(
                ["ticker", "fundamental"],
                as_index=False
            )
            .tail(1)
        )

        latest = latest.copy()

        latest["date"] = rebalance_date

        output.append(latest)

    if not output:
        return pd.DataFrame()

    return pd.concat(
        output,
        ignore_index=True
    )


def point_in_time_revenue_growth(
    fundamentals: pd.DataFrame,
    rebalance_dates: pd.DatetimeIndex
) -> pd.DataFrame:
    """Calculate point-in-time year-over-year growth from SEC quarter facts."""

    revenue = fundamentals.loc[
        fundamentals["fundamental"] == "revenue"
    ].copy()

    frame_parts = revenue["frame"].astype("string").str.extract(
        r"^CY(?P<year>\d{4})Q(?P<quarter>[1-4])$"
    )
    revenue = revenue.loc[frame_parts["year"].notna()].copy()

    if revenue.empty:
        return pd.DataFrame(
            columns=["date", "ticker", "revenue_growth"]
        )

    revenue["quarter_key"] = (
        frame_parts.loc[revenue.index, "year"].astype(int) * 4
        + frame_parts.loc[revenue.index, "quarter"].astype(int)
    )
    revenue["value"] = pd.to_numeric(revenue["value"], errors="coerce")
    revenue = revenue.dropna(subset=["filed", "end", "value"])

    output = []

    for rebalance_date in rebalance_dates:
        available = revenue.loc[revenue["filed"] <= rebalance_date]

        if available.empty:
            continue

        latest_by_quarter = (
            available
            .sort_values(["filed", "end"])
            .drop_duplicates(["ticker", "quarter_key"], keep="last")
        )
        latest_quarters = (
            latest_by_quarter
            .sort_values(["ticker", "quarter_key"])
            .groupby("ticker")
            .tail(1)
            .copy()
        )
        quarter_values = latest_by_quarter.set_index(
            ["ticker", "quarter_key"]
        )["value"]
        latest_quarters["prior_revenue"] = [
            quarter_values.get((ticker, quarter_key - 4), np.nan)
            for ticker, quarter_key in latest_quarters[
                ["ticker", "quarter_key"]
            ].itertuples(index=False, name=None)
        ]
        latest_quarters["revenue_growth"] = (
            latest_quarters["value"]
            / latest_quarters["prior_revenue"].replace(0, np.nan)
            - 1
        )
        latest_quarters["date"] = rebalance_date
        output.append(
            latest_quarters[["date", "ticker", "revenue_growth"]]
        )

    if not output:
        return pd.DataFrame(
            columns=["date", "ticker", "revenue_growth"]
        )

    return pd.concat(output, ignore_index=True)
rebalance_dates = prices.resample(
    CONFIG.rebalance_frequency
).last().index

pit_fundamentals = point_in_time_fundamentals(
    fundamentals,
    rebalance_dates
)

print(pit_fundamentals.shape)
# ============================================================
# CELL 8 — WIDE FUNDAMENTAL PANEL
# ============================================================

fundamental_panel = (
    pit_fundamentals
    .pivot_table(
        index=["date", "ticker"],
        columns="fundamental",
        values="value",
        aggfunc="last"
    )
    .reset_index()
)

fundamental_panel.columns.name = None

revenue_growth_panel = point_in_time_revenue_growth(
    fundamentals,
    rebalance_dates
)
fundamental_panel = fundamental_panel.merge(
    revenue_growth_panel,
    on=["date", "ticker"],
    how="left"
)


fundamental_panel.head()
# ============================================================
# CELL 9 — PRICE FACTOR ENGINEERING
# ============================================================

def calculate_price_features(
    prices: pd.DataFrame
) -> pd.DataFrame:
    """
    Generate cross-sectional equity features.

    Features:
        momentum_12_1
        momentum_6_1
        volatility_252
    """

    records = []

    for ticker in prices.columns:

        series = prices[ticker].dropna()

        if len(series) < CONFIG.min_history_days:
            continue

        df = pd.DataFrame(index=series.index)

        df["ticker"] = ticker
        df["price"] = series



        df["momentum_12_1"] = (
            series.shift(21)
            / series.shift(252)
            - 1
        )


        df["momentum_6_1"] = (
            series.shift(21)
            / series.shift(126)
            - 1
        )



        daily_returns = series.pct_change()

        df["volatility_252"] = (
            daily_returns
            .rolling(252)
            .std()
            * np.sqrt(252)
        )



        df["return_21d"] = (
            series / series.shift(21) - 1
        )

        records.append(df.reset_index())

    if not records:
        raise RuntimeError(
            "Unable to create price features."
        )

    output = pd.concat(
        records,
        ignore_index=True
    )

    output.rename(
        columns={"Date": "date"},
        inplace=True
    )

    return output


price_features = calculate_price_features(prices)

price_features.head()
# ============================================================
# CELL 10 — COMBINE DATASETS
# ============================================================


price_features["date"] = pd.to_datetime(
    price_features["date"]
)

fundamental_panel["date"] = pd.to_datetime(
    fundamental_panel["date"]
)

dataset = pd.merge_asof(
    price_features.sort_values(["date", "ticker"]),
    fundamental_panel.sort_values(["date", "ticker"]),
    on="date",
    by="ticker",
    direction="backward"
)

print(dataset.shape)

dataset.tail()
# ============================================================
# CELL 11 — MARKET CAPITALIZATION
# ============================================================

def add_market_cap(
    df: pd.DataFrame
) -> pd.DataFrame:
    """
    Estimate market capitalization.

    SEC shares outstanding are point-in-time fundamentals.
    Prices are observed on the rebalance date.

    market_cap = price × shares_outstanding
    """

    df = df.copy()

    if "shares" in df.columns:

        df["market_cap"] = (
            df["price"]
            * df["shares"]
        )

    else:


        df["market_cap"] = np.nan

    return df


dataset = add_market_cap(dataset)

dataset[
    ["date", "ticker", "price", "shares", "market_cap"]
].tail()
# ============================================================
# CELL 12 — FUNDAMENTAL FEATURE ENGINEERING
# ============================================================

def safe_divide(
    numerator: pd.Series,
    denominator: pd.Series
) -> pd.Series:

    denominator = denominator.replace(
        0,
        np.nan
    )

    result = numerator / denominator

    result = result.replace(
        [np.inf, -np.inf],
        np.nan
    )

    return result


def engineer_fundamental_features(
    df: pd.DataFrame
) -> pd.DataFrame:

    df = df.copy()


    df["earnings_yield"] = safe_divide(
        df["net_income"],
        df["market_cap"]
    )



    df["book_to_market"] = safe_divide(
        df["equity"],
        df["market_cap"]
    )



    df["roe"] = safe_divide(
        df["net_income"],
        df["equity"]
    )



    df["roa"] = safe_divide(
        df["net_income"],
        df["assets"]
    )



    df["operating_margin"] = safe_divide(
        df["operating_income"],
        df["revenue"]
    )


    df = df.sort_values(
        ["ticker", "date"]
    )

    return df


dataset = engineer_fundamental_features(
    dataset
)

dataset[
    [
        "date",
        "ticker",
        "earnings_yield",
        "book_to_market",
        "roe",
        "roa",
        "operating_margin",
        "revenue_growth"
    ]
].tail()
# ============================================================
# CELL 13 — ROBUST DATA CLEANING
# ============================================================

FACTOR_COLUMNS = [
    "momentum_12_1",
    "earnings_yield",
    "book_to_market",
    "roe",
    "revenue_growth",
    "volatility_252",
    "market_cap"
]


def winsorize_cross_section(
    series: pd.Series,
    lower: float = 0.01,
    upper: float = 0.99
) -> pd.Series:

    if series.dropna().empty:
        return series

    low = series.quantile(lower)
    high = series.quantile(upper)

    return series.clip(
        lower=low,
        upper=high
    )


def clean_features(
    df: pd.DataFrame
) -> pd.DataFrame:

    df = df.copy()

    for column in FACTOR_COLUMNS:

        if column not in df.columns:
            continue

        df[column] = (
            df.groupby("date")[column]
            .transform(
                lambda x: winsorize_cross_section(
                    x,
                    CONFIG.winsor_lower,
                    CONFIG.winsor_upper
                )
            )
        )

    return df


dataset = clean_features(dataset)

print(
    dataset[
        FACTOR_COLUMNS
    ].isna().mean().sort_values()
)
# ============================================================
# CELL 14 — CROSS-SECTIONAL STANDARDIZATION
# ============================================================

def cross_sectional_zscore(
    series: pd.Series
) -> pd.Series:

    std = series.std()

    if pd.isna(std) or std == 0:
        return pd.Series(
            0.0,
            index=series.index
        )

    return (
        series - series.mean()
    ) / std


def add_factor_scores(
    df: pd.DataFrame
) -> pd.DataFrame:

    df = df.copy()


    df["factor_momentum"] = (
        df.groupby("date")["momentum_12_1"]
        .transform(cross_sectional_zscore)
    )


    value_raw = (
        df["earnings_yield"].fillna(0)
        + df["book_to_market"].fillna(0)
    )

    df["factor_value"] = (
        value_raw.groupby(df["date"])
        .transform(cross_sectional_zscore)
    )


    quality_raw = (
        df["roe"].fillna(0)
        + df["roa"].fillna(0)
        + df["operating_margin"].fillna(0)
    )

    df["factor_quality"] = (
        quality_raw.groupby(df["date"])
        .transform(cross_sectional_zscore)
    )


    df["factor_growth"] = (
        df.groupby("date")["revenue_growth"]
        .transform(cross_sectional_zscore)
    )


    df["factor_low_vol"] = (
        -df.groupby("date")["volatility_252"]
        .transform(cross_sectional_zscore)
    )


    df["log_market_cap"] = np.log(
        df["market_cap"].where(
            df["market_cap"] > 0
        )
    )

    df["factor_size"] = (
        -df.groupby("date")["log_market_cap"]
        .transform(cross_sectional_zscore)
    )

    return df


dataset = add_factor_scores(dataset)

factor_columns = [
    "factor_momentum",
    "factor_value",
    "factor_quality",
    "factor_growth",
    "factor_low_vol",
    "factor_size"
]

dataset[
    ["date", "ticker"] + factor_columns
].tail()
# ============================================================
# CELL 15 — COMPOSITE ALPHA MODEL
# ============================================================

def calculate_composite_score(
    df: pd.DataFrame,
    weights: Dict[str, float]
) -> pd.DataFrame:

    df = df.copy()

    df["alpha_score"] = 0.0

    for factor_name, weight in weights.items():

        column = f"factor_{factor_name}"

        if column not in df.columns:
            raise KeyError(
                f"Missing factor column: {column}"
            )

        df["alpha_score"] += (
            weight * df[column].fillna(0)
        )


    df["alpha_percentile"] = (
        df.groupby("date")["alpha_score"]
        .rank(pct=True)
    )


    df["alpha_rank"] = (
        df.groupby("date")["alpha_score"]
        .rank(
            ascending=False,
            method="first"
        )
    )

    return df


dataset = calculate_composite_score(
    dataset,
    FACTOR_WEIGHTS
)

dataset[
    [
        "date",
        "ticker",
        "alpha_score",
        "alpha_percentile",
        "alpha_rank"
    ]
].tail(20)


# ============================================================
# CELL 16 — FORWARD RETURN CALCULATION
# ============================================================

def calculate_forward_returns(
    prices: pd.DataFrame,
    horizon: int = 21
) -> pd.DataFrame:

    forward_returns = (
        prices
        .shift(-horizon)
        / prices
        - 1
    )

    output = (
        forward_returns
        .stack()
        .rename("forward_return")
        .reset_index()
    )

    output.columns = [
        "date",
        "ticker",
        "forward_return"
    ]

    return output


forward_returns = calculate_forward_returns(
    prices,
    horizon=21
)

dataset = dataset.merge(
    forward_returns,
    on=["date", "ticker"],
    how="left"
)

dataset.tail()



# ============================================================
# CELL 17 — FACTOR INFORMATION COEFFICIENT
# ============================================================

def calculate_daily_ic(
    df: pd.DataFrame,
    factor: str,
    target: str = "forward_return"
) -> pd.Series:

    def daily_spearman(group):

        valid = group[
            [factor, target]
        ].dropna()

        if len(valid) < 5:
            return np.nan

        return spearmanr(
            valid[factor],
            valid[target]
        ).statistic

    return (
        df.groupby("date")
        .apply(
            daily_spearman,
            include_groups=False
        )
    )


ic_series = calculate_daily_ic(
    dataset,
    "alpha_score"
)

ic_mean = ic_series.mean()
ic_std = ic_series.std()

icir = (
    ic_mean / ic_std
    if ic_std > 0
    else np.nan
)

print(f"Mean IC : {ic_mean:.4f}")
print(f"ICIR    : {icir:.4f}")

# ============================================================
# CELL 18 — PORTFOLIO CONSTRUCTION
# ============================================================

def construct_long_only_portfolio(
    df: pd.DataFrame,
    top_quantile: float = 0.20
) -> pd.DataFrame:
    """
    Equal-weight the highest-ranked securities.

    Portfolio weights are calculated cross-sectionally
    at each rebalance date.
    """

    df = df.copy()

    df["selected"] = (
        df["alpha_percentile"]
        >= 1 - top_quantile
    )

    df["weight"] = 0.0

    for date, group in df.groupby("date"):

        selected = group["selected"]

        count = selected.sum()

        if count > 0:

            df.loc[
                group.index,
                "weight"
            ] = np.where(
                selected,
                1.0 / count,
                0.0
            )

    return df


portfolio = construct_long_only_portfolio(
    dataset,
    CONFIG.top_quantile
)

portfolio[
    [
        "date",
        "ticker",
        "alpha_score",
        "selected",
        "weight"
    ]
].tail(20)

# ============================================================
# CELL 19 — PORTFOLIO BACKTEST ENGINE
# ============================================================

def backtest_portfolio(
    portfolio: pd.DataFrame,
    prices: pd.DataFrame,
    transaction_cost_bps: float = 10
) -> pd.DataFrame:

    weights = (
        portfolio
        .pivot_table(
            index="date",
            columns="ticker",
            values="weight",
            aggfunc="last"
        )
        .fillna(0)
    )


    weights = (
        weights
        .reindex(prices.index)
        .ffill()
        .fillna(0)
    )

    returns = prices.pct_change()


    lagged_weights = weights.shift(1).fillna(0)

    gross_returns = (
        lagged_weights
        * returns
    ).sum(axis=1)


    turnover = (
        weights.diff()
        .abs()
        .sum(axis=1)
        .fillna(0)
    )


    transaction_cost = (
        turnover
        * transaction_cost_bps
        / 10000
    )

    net_returns = (
        gross_returns
        - transaction_cost
    )

    result = pd.DataFrame({
        "gross_return": gross_returns,
        "transaction_cost": transaction_cost,
        "net_return": net_returns,
        "turnover": turnover
    })

    result["nav"] = (
        1 + result["net_return"]
    ).cumprod()

    result["benchmark"] = (
        1 + returns.mean(axis=1).fillna(0)
    ).cumprod()

    return result


backtest = backtest_portfolio(
    portfolio,
    prices,
    CONFIG.transaction_cost_bps
)

backtest.tail()

# ============================================================
# CELL 20 — PERFORMANCE METRICS
# ============================================================

def performance_metrics(
    returns: pd.Series
) -> pd.Series:

    returns = returns.dropna()

    if len(returns) < 2:
        return pd.Series(dtype=float)

    periods = 252

    total_return = (
        (1 + returns).prod() - 1
    )

    years = len(returns) / periods

    cagr = (
        (1 + total_return) ** (1 / years) - 1
        if years > 0
        else np.nan
    )

    annual_vol = (
        returns.std() * np.sqrt(periods)
    )

    sharpe = (
        returns.mean() / returns.std()
        * np.sqrt(periods)
        if returns.std() > 0
        else np.nan
    )

    downside = returns[
        returns < 0
    ].std() * np.sqrt(periods)

    sortino = (
        returns.mean() * periods
        / downside
        if downside > 0
        else np.nan
    )

    wealth = (
        1 + returns
    ).cumprod()

    running_max = wealth.cummax()

    drawdown = (
        wealth / running_max - 1
    )

    max_drawdown = drawdown.min()

    calmar = (
        cagr / abs(max_drawdown)
        if max_drawdown != 0
        else np.nan
    )

    return pd.Series({
        "Total Return": total_return,
        "CAGR": cagr,
        "Annual Volatility": annual_vol,
        "Sharpe Ratio": sharpe,
        "Sortino Ratio": sortino,
        "Maximum Drawdown": max_drawdown,
        "Calmar Ratio": calmar,
        "Average Daily Turnover": returns.index.to_series().map(
            backtest["turnover"]
        ).mean()
    })


metrics = performance_metrics(
    backtest["net_return"]
)

metrics

# ============================================================
# CELL 21 — PERFORMANCE DASHBOARD
# ============================================================

plt.style.use("seaborn-v0_8-whitegrid")

fig, axes = plt.subplots(
    3,
    1,
    figsize=(14, 14),
    sharex=True
)


axes[0].plot(
    backtest.index,
    backtest["nav"],
    label="Multi-Factor Strategy",
    linewidth=2
)

axes[0].plot(
    backtest.index,
    backtest["benchmark"],
    label="Equal-Weight Benchmark",
    linewidth=1.5
)

axes[0].set_title(
    "Multi-Factor Equity Alpha — Cumulative Performance",
    fontsize=15,
    fontweight="bold"
)

axes[0].set_ylabel("Growth of $1")
axes[0].legend()


wealth = backtest["nav"]

drawdown = (
    wealth / wealth.cummax() - 1
)

axes[1].fill_between(
    drawdown.index,
    drawdown.values,
    0,
    alpha=0.30
)

axes[1].set_title(
    "Strategy Drawdown"
)

axes[1].set_ylabel("Drawdown")


axes[2].plot(
    backtest.index,
    backtest["turnover"],
    linewidth=1
)

axes[2].set_title(
    "Portfolio Turnover"
)

axes[2].set_ylabel("Turnover")

plt.tight_layout()

plt.savefig(
    CHART_DIR / "performance_dashboard.png",
    dpi=200,
    bbox_inches="tight"
)

plt.show()

# ============================================================
# CELL 22 — FACTOR CORRELATION MATRIX
# ============================================================

factor_matrix = dataset[factor_columns].corr()

plt.figure(
    figsize=(10, 8)
)

sns.heatmap(
    factor_matrix,
    annot=True,
    fmt=".2f",
    center=0,
    square=True
)

plt.title(
    "Factor Correlation Matrix",
    fontsize=15,
    fontweight="bold"
)

plt.tight_layout()

plt.savefig(
    CHART_DIR / "factor_correlation.png",
    dpi=200,
    bbox_inches="tight"
)

plt.show()

# ============================================================
# CELL 23 — FACTOR QUINTILE ANALYSIS
# ============================================================

def factor_quintile_returns(
    df: pd.DataFrame,
    factor: str
) -> pd.DataFrame:

    temp = df[
        [
            "date",
            "ticker",
            factor,
            "forward_return"
        ]
    ].dropna()

    temp["quintile"] = (
        temp.groupby("date")[factor]
        .transform(
            lambda x: pd.qcut(
                x.rank(method="first"),
                5,
                labels=False
            ) + 1
        )
    )

    result = (
        temp
        .groupby(
            ["date", "quintile"]
        )["forward_return"]
        .mean()
        .unstack()
    )

    return result


momentum_quintiles = factor_quintile_returns(
    dataset,
    "factor_momentum"
)

cumulative = (
    1 + momentum_quintiles
).cumprod()

plt.figure(
    figsize=(14, 7)
)

for column in cumulative.columns:

    plt.plot(
        cumulative.index,
        cumulative[column],
        label=f"Q{column}"
    )

plt.title(
    "Momentum Factor — Quintile Performance",
    fontsize=15,
    fontweight="bold"
)

plt.ylabel("Growth of $1")
plt.legend()

plt.tight_layout()

plt.savefig(
    CHART_DIR / "momentum_quintiles.png",
    dpi=200,
    bbox_inches="tight"
)

plt.show()

# ============================================================
# CELL 24 — INFORMATION COEFFICIENT DASHBOARD
# ============================================================

fig, axes = plt.subplots(
    2,
    1,
    figsize=(14, 9)
)


rolling_ic = ic_series.rolling(
    63
).mean()

axes[0].plot(
    rolling_ic.index,
    rolling_ic,
    linewidth=2
)

axes[0].axhline(
    0,
    linestyle="--",
    linewidth=1
)

axes[0].set_title(
    "63-Day Rolling Information Coefficient",
    fontsize=14,
    fontweight="bold"
)

axes[0].set_ylabel("IC")


axes[1].hist(
    ic_series.dropna(),
    bins=30
)

axes[1].axvline(
    ic_mean,
    linestyle="--",
    linewidth=2,
    label=f"Mean IC = {ic_mean:.3f}"
)

axes[1].set_title(
    "Information Coefficient Distribution"
)

axes[1].set_xlabel("Spearman IC")
axes[1].legend()

plt.tight_layout()

plt.savefig(
    CHART_DIR / "information_coefficient.png",
    dpi=200,
    bbox_inches="tight"
)

plt.show()

# ============================================================
# CELL 25 — CURRENT PORTFOLIO
# ============================================================

latest_date = portfolio["date"].max()

current_portfolio = (
    portfolio[
        portfolio["date"] == latest_date
    ]
    [
        [
            "ticker",
            "alpha_score",
            "alpha_percentile",
            "weight"
        ]
    ]
    .sort_values(
        "weight",
        ascending=False
    )
)

current_portfolio

# ============================================================
# CELL 26 — PORTFOLIO FACTOR EXPOSURE
# ============================================================

latest = portfolio[
    portfolio["date"] == latest_date
].copy()

factor_exposure = {}

for factor in factor_columns:

    factor_exposure[factor] = (
        latest[factor]
        * latest["weight"]
    ).sum()

factor_exposure = (
    pd.Series(factor_exposure)
    .sort_values()
)

plt.figure(
    figsize=(10, 6)
)

factor_exposure.plot(
    kind="barh"
)

plt.axvline(
    0,
    linestyle="--",
    linewidth=1
)

plt.title(
    "Current Portfolio Factor Exposure",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("Weighted Factor Exposure")

plt.tight_layout()

plt.savefig(
    CHART_DIR / "factor_exposure.png",
    dpi=200,
    bbox_inches="tight"
)

plt.show()

# ============================================================
# CELL 27 — SAVE RESEARCH OUTPUTS
# ============================================================


dataset.to_parquet(
    PROCESSED_DIR / "factor_dataset.parquet",
    index=False
)

portfolio.to_parquet(
    PROCESSED_DIR / "portfolio_weights.parquet",
    index=False
)

backtest.to_parquet(
    PROCESSED_DIR / "backtest_results.parquet",
    index=True
)

metrics.to_csv(
    TABLE_DIR / "performance_metrics.csv"
)

current_portfolio.to_csv(
    TABLE_DIR / "current_portfolio.csv",
    index=False
)

factor_exposure.to_csv(
    TABLE_DIR / "factor_exposure.csv"
)

print("Research outputs saved.")

# ============================================================
# CELL 28 — FINAL RESEARCH SUMMARY
# ============================================================

print("=" * 70)
print("INSTITUTIONAL MULTI-FACTOR EQUITY ALPHA MODEL")
print("=" * 70)

print("\nUNIVERSE")
print("-" * 70)
print(f"Securities:       {len(TICKERS)}")
print(f"Start date:       {prices.index.min().date()}")
print(f"End date:         {prices.index.max().date()}")

print("\nFACTOR WEIGHTS")
print("-" * 70)

for factor, weight in FACTOR_WEIGHTS.items():
    print(
        f"{factor:<15} {weight:.1%}"
    )

print("\nPERFORMANCE")
print("-" * 70)

for metric, value in metrics.items():

    if isinstance(value, (float, np.floating)):

        if "Ratio" in metric:
            print(
                f"{metric:<25} {value:.2f}"
            )
        else:
            print(
                f"{metric:<25} {value:.2%}"
            )

print("\nALPHA QUALITY")
print("-" * 70)

print(
    f"Mean IC:          {ic_mean:.4f}"
)

print(
    f"ICIR:             {icir:.4f}"
)

print("\nLatest rebalance:")
print(latest_date)

print("\nTop holdings:")

display(
    current_portfolio.head(10)
)

print("\nResearch complete.")
```
